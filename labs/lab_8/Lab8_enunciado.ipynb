## **MLFlow**

`MLflow` es una plataforma de código abierto que simplifica la gestión y seguimiento de proyectos de aprendizaje automático. Con sus herramientas, los desarrolladores pueden organizar, rastrear y comparar experimentos, además de registrar modelos y controlar versiones.

<p align="center">
  <img src="https://spark.apache.org/images/mlflow-logo.png" width="350">
</p>

Si bien esta plataforma cuenta con un gran número de herramientas y funcionalidades, en este laboratorio trabajaremos con dos:
1. **Runs**: Registro que constituye la información guardada tras la ejecución de un entrenamiento. Cada `run` tiene su propio run_id, el cual sirve como identificador para el entrenamiento en sí mismo. Dentro de cada `run` podremos acceder a información como los hiperparámetros utilizados, las métricas obtenidas, las librerías requeridas y hasta nos permite descargar el modelo entrenado.
2. **Experiments**: Se utilizan para agrupar y organizar diferentes ejecuciones de modelos (`runs`). En ese sentido, un experimento puede agrupar 1 o más `runs`. De esta manera, es posible también registrar métricas, parámetros y archivos (artefactos) asociados a cada experimento.

### **Todo bien pero entonces, ¿cómo se usa en la práctica `MLflow`?**

Es sencillo! Considerando un problema de machine learning genérico, podemos registrar la información relevante del entrenamiento ejecutando `mlflow.autolog()` antes entrenar nuestro modelo. Veamos este bonito ejemplo facilitado por los mismos creadores de `MLflow`:

```python
#!pip install mlflow
import mlflow # importar mlflow

from sklearn.model_selection import train_test_split
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor

db = load_diabetes()
X_train, X_test, y_train, y_test = train_test_split(db.data, db.target)

# Create and train models.
rf = RandomForestRegressor(n_estimators=100, max_depth=6, max_features=3)

mlflow.autolog() # registrar automáticamente información del entrenamiento
with mlflow.start_run(): # delimita inicio y fin del run
    # aquí comienza el run
    rf.fit(X_train, y_train) # train the model
    predictions = rf.predict(X_test) # Use the model to make predictions on the test dataset.
    # aquí termina el run
```

Si ustedes ejecutan el código anterior en sus máquinas locales (desde un jupyter notebook por ejemplo) se darán cuenta que en su directorio *root* se ha creado la carpeta `mlruns`. Esta carpeta lleva el tracking de todos los entrenamientos ejecutados desde el directorio root (importante: si se cambian de directorio y vuelven a ejecutar el código anterior, se creará otra carpeta y no tendrán acceso al entrenamiento anterior). Para visualizar estos entrenamientos, `MLflow` nos facilita hermosa interfaz visual a la que podemos acceder ejecutando:


````python
import mlflow  # importar mlflow
from sklearn.datasets import load_diabetes
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

db = load_diabetes()
X_train, X_test, y_train, y_test = train_test_split(db.data, db.target)

# Create and train models.
rf = RandomForestRegressor(n_estimators=100, max_depth=6, max_features=3)

mlflow.autolog()  # registrar automáticamente información del entrenamiento
with mlflow.start_run():  # delimita inicio y fin del run
    # aquí comienza el run
    rf.fit(X_train, y_train)  # train the model
    predictions = rf.predict(X_test)  # Use the model to make predictions on the test dataset.
    # aquí termina el run
````

```python
def get_best_model(experiment_id):
    runs = mlflow.search_runs(experiment_id)
```

## **3.2 Preguntas de Smapina (1 punto)**
Tras haber experimentado con Docker, Smapina desea profundizar más en el tema y decide realizarle las siguientes consultas:

**¿Cómo se diferencia Docker de una máquina virtual (VM)?**

Una máquina virtual virtualiza hardware completo e incluye su propio sistema operativo invitado, lo que la hace pesada en consumo de RAM y disco y lenta en arranque. Docker, en cambio, utiliza contenedores que comparten el kernel del sistema operativo anfitrión, solo empaqueta la aplicación y sus dependencias a nivel de espacio de usuario. Esto los hace mucho más livianos, rápidos de iniciar y con menor uso de CPU y memoria.

**¿Cuál es la diferencia entre usar Docker y ejecutar la aplicación directamente en el sistema local?**

Ejecutar la aplicación directamente en el sistema local depende del entorno específico de la máquina. Versión de Python, librerías instaladas globalmente, variables de entorno y configuraciones del SO, etc. Esto genera el tipico problema de funciona en mi máquina, pero no en la de los demás. Docker empaqueta la aplicación junto con todas sus dependencias en una imagen inmutable, garantizando que el mismo artefacto se ejecute de forma idéntica en cualquier máquina que tenga Docker instalado. Además, el contenedor está aislado del sistema anfitrión, por lo que no interfiere con otras aplicaciones ni con el estado global del sistema.

**¿Cómo asegura Docker la consistencia entre diferentes entornos de desarrollo y producción?**

Docker asegura la consistencia mediante el uso de imágenes. Artefactos inmutables que contienen el sistema de archivos completo de la aplicación. Al construir la imagen una sola vez a partir del `Dockerfile` y después distribuirla. Logra eliminar discrepancias de versiones de librerías o del intérprete entre entornos. Además, las variables de entorno específicas de cada entorno se inyectan en tiempo de ejecución mediante flags `-e` o archivos `.env`, manteniendo la imagen genérica y el comportamiento reproducible.

**¿Cómo se gestionan los volúmenes en Docker para la persistencia de datos?**

Por defecto, el sistema de archivos de un contenedor es efímero. Cualquier dato escrito dentro de él se pierde al detener o eliminar el contenedor. Los volúmenes de Docker resuelven esto montando un directorio del host dentro del contenedor. Existen dos mecanismos principales: los **bind mounts** (`-v /ruta/host:/ruta/contenedor`), que mapean un directorio específico del anfitrión, y los **named volumes** (`docker volume create nombre`), gestionados por Docker y almacenados en su directorio interno.

**¿Qué son Dockerfile y docker-compose.yml, y cuál es su propósito?**

Un **Dockerfile** es un archivo de texto con instrucciones paso a paso para construir una imagen Docker. Define la imagen base (`FROM`), copia archivos (`COPY`), instala dependencias (`RUN`), expone puertos (`EXPOSE`) y especifica el comando de inicio (`CMD`).

**docker-compose.yml** es un archivo de configuración declarativo que permite definir y orquestar múltiples contenedores como un único servicio. `docker-compose.yml` describe todos los servicios, por ejemplo, la API FastAPI, una base de datos PostgreSQL y un servidor MLflow, sus dependencias, volúmenes, redes y variables de entorno en un solo lugar. Con un `docker compose up` se levanta toda la infraestructura de un Docker multicontenedor.
